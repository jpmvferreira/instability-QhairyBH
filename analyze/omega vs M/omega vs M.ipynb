```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# display dynamical plots
%matplotlib widget
```

```python
fig, ax1 = plt.subplots(1, 1, figsize=(11.25,5))

rh_list = ["0,2", "0,25", "0,3"]

for rh in rh_list:
    omega_list = []
    M_list = []

    file = f"data/rh={rh}.txt"
    with open(file) as file:
        for line in file:
            omega, M = line.split()
            omega_list.append(float(omega))
            M_list.append(float(M))

    ax1.plot(omega_list, M_list, label=r"$r_h=$"+rh.replace(",","."))

ax1.text(0.955, 0.5,  r"$r_h=0.2$",  fontsize=14)
ax1.text(0.955, 0.61, r"$r_h=0.25$", fontsize=14)
ax1.text(0.955, 0.73, r"$r_h=0.3$",  fontsize=14)

dots = np.array([
    [0.95, 1.19118],
    [0.9, 1.01713],
    [0.85, 0.86989],
    [0.8, 0.745411],
    [0.75, 0.638951],
    [0.7, 0.539687],
    [0.7, 0.473189],
    [0.75, 0.46182],
    [0.8, 0.461563],
    [0.85, 0.46457],
    [0.9, 0.469372],
    [0.95, 0.475467]
])
ax1.scatter(dots[:,0], dots[:,1], color="C0", zorder=10)

dots = np.array([
    [0.95, 1.29556],
    [0.95, 0.59265],
    [0.9,  1.12069],
])
ax1.scatter(dots[:,0], dots[:,1], color="C1", zorder=10)

dots = np.array([
    [0.95, 1.39999],
    [0.95, 0.70991],
    [0.9,  1.22429],
    [0.9,  0.700242],
    [0.85, 1.07538],
    [0.85, 0.692208],
    [0.8,  0.949066],
    [0.8,  0.686475],
    [0.75, 0.840005],
    [0.75, 0.685122],
    [0.7,  0.719402]
])
ax1.scatter(dots[:,0], dots[:,1], color="C2", zorder=10)

ax1.grid(alpha=0.5)
ax1.set_xlim(left=0.675, right=1)
ax1.set_xlabel(r"$\omega$", fontsize=18)
ax1.set_ylabel(r"$M_T$", fontsize=18);
#ax1.legend();
```
